### Logistic Regression
Logistic regression is a statistical method for analyzing a dataset in which there are one or more independent variables that determine an outcome. The outcome is typically a binary variable (0 or 1, true or false, yes or no). The logistic regression model estimates the probability that a given input point belongs to a particular category. The model uses the logistic function to map predicted values to probabilities, which can then be used for classification. The logistic regression equation is given by:
The sigmoid function is defined as:
$$\sigma(z) = \frac{1}{1 + e^{-z}}
$$Where:
- $z$ is the linear combination of the input features and their corresponding weights, plus the bias term. It can be expressed as:
$$z = w_1x_1 + w_2x_2 + ... + w_nx_n + b
$$Where:- $w_i$ are the weights associated with each input feature $x_i$.
- $b$ is the bias term.
The logistic regression model can be trained using maximum likelihood estimation, whichaims to find the parameters (weights and bias) that maximize the likelihood of the observed data given the model. The cost function used for logistic regression is the binary cross-entropy loss, which is defined as:
$$J(w, b) = -\frac{1}{m} \sum_{i=1}^{m} \left[ y^{(i)} \log(\hat{y}^{(i)}) + (1 - y^{(i)}) \log(1 - \hat{y}^{(i)}) \right]
$$Where:- $m$ is the number of training examples.
- $y^{(i)}$ is the true label for the $i$-th training example.
- $\hat{y}^{(i)}$ is the predicted probability for the $i$-th training example, calculated using the logistic function.
The parameters can be optimized using gradient descent or other optimization algorithms to minimize the cost function and improve the model's performance. Logistic regression is widely used in various applications, including binary classification problems such asspam detection, medical diagnosis, and customer churn prediction.

```python
import pandas as pd
import numpy as np
```

```python
# Read the data
data = {"Study_Hours": [2, 3, 4, 5, 6, 7, 8, 1, 9, 10, 3,6],
        "Mock_Test_Score": [4.5, 5.0, 5.5, 6.0, 6.5, 7.0, 7.5, 4.0, 8.0, 8.5, 4.8, 6.8],
        "Extra_Class": [0, 0, 1, 0, 1, 1, 0, 0, 1, 1, 1, 0],
        "Label": [0, 0, 1, 0, 1, 1, 1, 0, 1, 1, 0, 1]}

df = pd.DataFrame(data)
print(df)
```

```python
# Start at all zeroes
B0 = 0
B1 = 0
B2 = 0
B3 = 0
```

```python
# Iterate 10 times
for i in range(10):
    # Calculate the predicted probabilities
    z = B0 + B1 * df["Study_Hours"] + B2 * df["Mock_Test_Score"] + B3 * df["Extra_Class"]
    predicted_probabilities = 1 / (1 + np.exp(-z))
        
    # Calculate the gradients
    error = predicted_probabilities - df["Label"]
    dB0 = np.sum(error)
    dB1 = np.sum(error * df["Study_Hours"])
    dB2 = np.sum(error * df["Mock_Test_Score"])
    dB3 = np.sum(error * df["Extra_Class"])
        
    # Update the coefficients
    learning_rate = 0.01
    B0 -= learning_rate * dB0
    B1 -= learning_rate * dB1
    B2 -= learning_rate * dB2
    B3 -= learning_rate * dB3
    print(f"Iteration {i+1}: B0={B0:.4f}, B1={B1:.4f}, B2={B2:.4f}, B3={B3:.4f}")
```

```python
# Make predictions
z = B0 + B1 * df["Study_Hours"] + B2 * df["Mock_Test_Score"] + B3 * df["Extra_Class"]
predicted_probabilities = 1 / (1 + np.exp(-z))
predicted_labels = (predicted_probabilities >= 0.5).astype(int)
# Add it to the DataFrame
df["Predicted_Probabilities"] = predicted_probabilities
df["Predicted_Labels"] = predicted_labels
df["Actual_Labels"] = df["Label"]
print(df[['Predicted_Probabilities', 'Predicted_Labels', 'Actual_Labels']])
```

```python
Accuracy = np.mean(predicted_labels == df["Label"])
print(f"Accuracy: {Accuracy:.4f}")
```
