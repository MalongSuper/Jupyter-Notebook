### 1. Introduction to Linear Regression Formulas

Linear regression aims to model the relationship between a dependent variable ($Y$) and an independent variable ($X$) by fitting a linear equation to observed data.

The relationship is expressed as:

$$Y = A + BX$$

Where:
- $B$ (Slope): Represents the rate of change in $Y$ for every unit change in $X$.
- $A$ (Intercept): The value of $Y$ when $X = 0$.

To find the "line of best fit," we use the Ordinary Least Squares (OLS) method to calculate $A$ and $B$ using the following components derived from the dataset:

Slope ($B$):

$$B = \frac{n(\sum XY) - (\sum X)(\sum Y)}{n(\sum X^2) - (\sum X)^2}$$

Intercept ($A$):

$$A = \frac{\sum Y - (B \cdot \sum X)}{n}$$

### 2. Code Workflow

The implementation follows a structured mathematical approach using pandas for data manipulation.

**Step 1: Environment Setup and Data Loading**

The process begins by importing pandas and defining a sample dataset consisting of two variables, $X$ and $Y$, which are then stored in a DataFrame.

```python
import pandas as pd
```

```python
sample_data = {
    'X': [4.5, 5.2, 6.0, 6.4, 7.1, 7.3, 8.0, 8.4, 9.1, 9.5,
          4.8, 5.5, 5.9, 6.6, 6.9, 7.5, 8.1, 8.6, 8.9, 9.3],
    'Y': [5.1, 6.3, 6.7, 7.5, 7.8, 8.1, 8.6, 9.0, 9.4, 9.8,
          5.4, 6.1, 6.5, 7.2, 7.6, 8.0, 8.3, 8.7, 9.1, 9.5]
}

df = pd.DataFrame(sample_data, columns=['X', 'Y'])
```

**Step 2: Statistical Component Calculation**

Before applying the regression formula, several key statistical values are calculated from the DataFrame:

- $n$: The total number of samples.
- $\sum X$ and $\sum Y$: The total sums of the independent and dependent variables.
- $\sum XY$: The sum of the product of each $X$ and $Y$ pair.
- $\sum X^2$: The sum of the squares of each $X$ value.
- $(\sum X)^2$: The square of the total sum of $X$.

```python
n = len(df)
sum_x = df['X'].sum()
sum_y = df['Y'].sum()
sum_xy = (df['X'] * df['Y']).sum()
sum_x_pow = (df['X'] ** 2).sum()
sum_pow_x = (df['X'].sum()) ** 2

print(f"n = {n}")
print(f"Sum of X = {sum_x}")
print(f"Sum of Y = {sum_y}")
print(f"Sum of XY = {sum_xy}")
print(f"Sum of X^2 = {sum_x_pow}")
print(f"(Sum of X)^2 = {sum_pow_x}")
```

**Step 3: Parameter Estimation**

Using the formulas defined in the introduction, the code calculates the specific values for $B$ (slope) and $A$ (intercept).  This results in the final regression equation

```python
B = ((n * sum_xy) - (sum_x * sum_y)) / ((n * sum_x_pow) - sum_pow_x)
A = (sum_y - (B * sum_x)) / n
print(f"Linear Regression: Y = {A} + {B}X")
```

**Step 4: Prediction and Visualization**

With the parameters estimated, the code can predict $Y$ values based on new $X$ inputs.

```python
pred_y = []
for i in range(len(df)):
    # Append the result to list
    pred_y.append(A + B * df['X'][i])

# Construct a new data frame
predict_data = {'Actual Y': df['Y'].tolist(), 'Predicted Y': pred_y}
pred_df = pd.DataFrame(predict_data)
# Calculate the residuals
pred_df['Residual'] = abs(pred_df['Actual Y'] - pred_df['Predicted Y'])
print(pred_df)
```

Then, we compute Sum of Squared Errors (SSE), Mean Squared Error (MSE), and R^2 to evaluate the model's performance.

```python
print(f"Sum of Squared Error: {(pred_df['Residual'] ** 2).sum()}")
print(f"Mean Squared Error: {(pred_df['Residual'] ** 2).sum() / len(pred_df)}")
# Calculate R^2
ss_res = (pred_df['Residual'] ** 2).sum()
ss_tot = abs((pred_df['Actual Y'] - pred_df['Actual Y'].mean()) ** 2).sum()
print(f"R^2 Score: {1 - (ss_res / ss_tot)}")
```

Additionally, the regression line can be visualized using libraries such as Matplotlib to show how well the model fits the data.

```python
import matplotlib.pyplot as plt

# Draw Regression Line
# Scatter plot of actual data
plt.scatter(df['X'], df['Y'],
            color='blue', label='Actual Data')
# Regression line
X_range = [min(df['X']), max(df['Y'])]
y_range = [A + B * x for x in X_range]
plt.plot(X_range, y_range, color='red', linewidth=2, label='Regression Line')

# Labels and title
plt.xlabel('X')
plt.ylabel('Y')
plt.title(f'Linear Regression: Y = {A:.2f} + ({B:.2f} * X)')
plt.legend()
plt.grid(True)
plt.show()
```

## Key Takeaways

- **Manual Calculation vs. Libraries**: While libraries like `scikit-learn` automate these processes, implementing the formulas manually demonstrates that regression is essentially a system of statistical summations.

- **Error Minimization**: The goal of the OLS method is to minimize the Sum of Squared Errors (SSE), ensuring the line is as close to all data points as possible.

- **Predictive Power**: The $R^2$ score is vital for understanding how much of the variance in $Y$ can be explained by $X$; a score closer to 1 indicates a stronger model.
