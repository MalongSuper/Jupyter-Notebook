## Vietnamese in Natural Language Processing (NLP) with Underthesea library

The underthesea library is a powerful tool for processing Vietnamese text in natural language processing (NLP) tasks. It provides various functionalities such as tokenization, part-of-speech tagging, named entity recognition, and more.

Reference: https://underthesea.readthedocs.io/en/latest/readme.html

```python
# Import the underthesea library for Vietnamese NLP tasks
import underthesea as ut
```

- **Sentence Segmentation**: This is the process of dividing a text into individual sentences. Underthesea can accurately segment Vietnamese text into sentences, which is crucial for many NLP applications.

```python
text = 'Taylor cho biết lúc đầu cô cảm thấy ngại với cô bạn thân Amanda nhưng rồi mọi thứ trôi qua nhanh chóng. Amanda cũng thoải mái với mối quan hệ này.'
# Tokenize the Vietnamese text into words
tokens = ut.word_tokenize(text)
print(tokens)
```

- **Text Normalization**: This is the process of converting text into a standard format. Underthesea can normalize Vietnamese text by removing diacritics, converting to lowercase, and more.

```python
text = 'Ðảm baỏ chất lựơng phòng thí nghịêm hoá học, sinh học và môi trường'
# Normalize the Vietnamese text by removing diacritics and converting to lowercase
normalized_text = ut.text_normalize(text)
print(normalized_text)
```

- **Tagging**: Word segmentation, POS tagging, chunking, and dependency parsing

```python
sentence = 'Chàng trai 9X Quảng Trị khởi nghiệp từ nấm sò'
# Perform word segmentation on the Vietnamese sentence
segmented_sentence = ut.word_tokenize(sentence)
print(segmented_sentence)
# Perform POS tagging on the Vietnamese sentence
pos_tags = ut.pos_tag(sentence)
print(pos_tags)
# Perform chunking on the Vietnamese sentence
chunks = ut.chunk(sentence)
print(chunks)
# Perform dependency parsing on the Vietnamese sentence
dependencies = ut.dependency_parse(sentence)
print(dependencies)
```

- **Named Entity Recognition**: The ability to identify and classify named entities in text, such as people, organizations, locations, etc.

```python
text = 'Chưa tiết lộ lịch trình tới Việt Nam của Tổng thống Mỹ Donald Trump'
# Perform named entity recognition on the Vietnamese text
entities = ut.ner(text)
print(entities)
```

- **Text classification and sentiment analysis**: Underthesea can be used for text classification tasks, including sentiment analysis, which is the process of determining the sentiment or emotion expressed in a piece of text.

```python
text1 = 'HLV đầu tiên ở Premier League bị sa thải sau 4 vòng đấu'
text2 = 'Hội đồng tư vấn kinh doanh Asean vinh danh giải thưởng quốc tế'
# Perform text classification on the Vietnamese texts
classification1 = ut.classify(text1)
classification2 = ut.classify(text2)
print(classification1)
print(classification2)
```

```python
text1 = 'hàng kém chất lg,chăn đắp lên dính lông lá khắp người. thất vọng'
text2 = 'sản phẩm tốt, giao hàng nhanh, đóng gói cẩn thận. rất hài lòng'
text3 = 'Sản phẩm hơi nhỏ so với tưởng tượng nhưng chất lượng tốt, đóng gói cẩn thận.'
# Perform sentiment analysis on the Vietnamese texts
sentiment1 = ut.sentiment(text1)
sentiment2 = ut.sentiment(text2)
sentiment3 = ut.sentiment(text3)
print(sentiment1)
print(sentiment2)
print(sentiment3)
```

- **Identifying the Language of Text**: Underthesea can identify the language of a given text, which is useful for multilingual applications.

```python
text = 'Cựu binh Mỹ trả nhật ký nhẹ lòng khi thấy cuộc sống hòa bình tại Việt Nam'
# Identify the language of the given text
language = ut.lang_detect(text)
print(language)
```

- **Translating Vietnamese text to English**: Underthesea can be used to translate Vietnamese text into English, which is helpful for applications that require cross-lingual understanding.

```python
text = 'Cựu binh Mỹ trả nhật ký nhẹ lòng khi thấy cuộc sống hòa bình tại Việt Nam'
# Translate the Vietnamese text to English
translation = ut.translate(text, target_lang='en')
print(translation)
```
