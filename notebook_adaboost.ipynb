### Adaptive Boosting

Adaptive Boosting (AdaBoost) is an ensemble machine learning algorithm that combines multiple "weak classifiers" (typically simple decision trees) in sequence to create a single, highly accurate "strong classifier". It improves performance iteratively by assigning higher weights to misclassified data points, forcing subsequent models to focus on hard-to-learn examples. 

Key aspects of the Adaptive Boosting algorithm include:
- Iterative Learning: The algorithm begins with a base model and, in each iteration, adapts by training a new model that focuses on the errors of the previous one.

- Weighted Data Points: Initially, all data points have equal weights. After each model trains, the weights for misclassified samples are increased, making them more significant for the next learner.

- Weak Learner Focus: While it often uses "decision stumps" (single-level decision trees) as base learners, any machine learning algorithm can be used.

- Final Prediction: The final model combines predictions from all weak learners using a weighted majority vote, where more accurate models have more influence.

- Applications & Limitations: It is widely used for classification (and sometimes regression) tasks, but it is sensitive to noisy data and outliers, which can lead to overfitting
